### 8-Puzzle Solver using DFS

This section implements a Depth-First Search (DFS) algorithm to solve the 8-puzzle problem. The goal is to transform an initial board configuration into a target configuration by sliding tiles into the empty space.

````python
from collections import deque

# Helper function to find the blank space (represented by '_')
def find_blank(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == '_':
                return r, c
    return -1, -1 # Should not happen in a valid 8-puzzle

# Helper function to get possible next states
def get_neighbors(state):
    neighbors = []
    r, c = find_blank(state)

    # Possible moves (up, down, left, right)
    moves = [
        (-1, 0),  # Up
        (1, 0),   # Down
        (0, -1),  # Left
        (0, 1)    # Right
    ]

    for dr, dc in moves:
        new_r, new_c = r + dr, c + dc

        if 0 <= new_r < 3 and 0 <= new_c < 3:
            # Create a new state by swapping the blank with the neighbor
            new_state = [list(row) for row in state]
            new_state[r][c], new_state[new_r][new_c] = new_state[new_r][new_c], new_state[r][c]
            neighbors.append(tuple(tuple(row) for row in new_state))

    return neighbors

# DFS function to solve the 8-puzzle with a depth limit
def solve_8_puzzle_dfs(initial_state, final_state, max_depth=None):
    initial_state_tuple = tuple(tuple(row) for row in initial_state)
    final_state_tuple = tuple(tuple(row) for row in final_state)

    # Stack for DFS (LIFO) - stores (current_state, path, current_depth)
    stack = [(initial_state_tuple, [initial_state_tuple], 0)]
    visited = set()
    states_explored = 0

    while stack:
        current_state, path, current_depth = stack.pop()

        states_explored += 1

        if current_state == final_state_tuple:
            return path, states_explored

        if current_state in visited:
            continue

        visited.add(current_state)

        if max_depth is not None and current_depth >= max_depth:
            continue

        for neighbor in reversed(get_neighbors(current_state)):
            if neighbor not in visited:
                stack.append((neighbor, path + [neighbor], current_depth + 1))

    return None, states_explored # No solution found within the depth limit or in general
````

````python
# Define the initial and final states as per the problem description
initial_state = [
    [2, 8, 3],
    [1, 6, 4],
    [7, '_', 5]
]

final_state = [
    [1, 2, 3],
    [8, '_', 4],
    [7, 6, 5]
]

print("Initial State:")
for row in initial_state:
    print(row)

print("\nFinal State:")
for row in final_state:
    print(row)

# Solve the puzzle with a depth limit of 6
solution_path, states_count = solve_8_puzzle_dfs(initial_state, final_state, max_depth=6)

if solution_path:
    print(f"\nSolution Found in {len(solution_path) - 1} steps, exploring {states_count} states (with depth limit 6):")
    for i, state in enumerate(solution_path):
        print(f"\nStep {i}:")
        for row in state:
            print(row)
else:
    print(f"\nNo solution found within depth limit 6 after exploring {states_count} states.")
````
