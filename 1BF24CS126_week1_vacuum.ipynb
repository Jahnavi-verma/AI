## Vaccum cleaner

```python
class VaccumCleaner:
    def __init__(self, location):
        self.location= location
    def act(self, environment):
      if environment[self.location]=="Dirty":
        print("Location", self.location, "is Dirty")
        environment[self.location]="Clean"
      else:
        if self.location == "A":
          print("Moving from A to B")
          self.location = "B"
        else:
          print("Moving from B to A")
          self.location = "A"
environment={
    "A":"Dirty",
    "B":"Dirty"
}
```

```python
agent = VaccumCleaner("A")
```

```python
print("Intial Environment:", environment)
for step in range(4):
  print(f"Step {step+1}: Agent at {agent.location}")
  agent.act(environment)
  print("Environment:", environment)
  print()
print(f"Final Environment: {environment}")
```
