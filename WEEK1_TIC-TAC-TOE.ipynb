```python
import random
import time

# Constants for the game
COMPUTER = 1
HUMAN = 2
SIDE = 3
COMPUTERMOVE = 'O'
HUMANMOVE = 'X'

def initialise():
    board = [[' ' for _ in range(SIDE)] for _ in range(SIDE)]
    moves = [i for i in range(SIDE*SIDE)]
    random.shuffle(moves)
    return board, moves

def showBoard(board):
    print("\n\n")
    print("\t\t\t {} | {} | {} ".format(board[0][0], board[0][1], board[0][2]))
    print("\t\t\t--------------")
    print("\t\t\t {} | {} | {} ".format(board[1][0], board[1][1], board[1][2]))
    print("\t\t\t--------------")
    print("\t\t\t {} | {} | {} \n\n".format(board[2][0], board[2][1], board[2][2]))

def showInstructions():
    print("\t\t\t Tic-Tac-Toe\n\n")
    print("Choose a cell numbered from 1 to 9 as below and play\n\n")
    print("\t\t\t 1 | 2 | 3 ")
    print("\t\t\t--------------")
    print("\t\t\t 4 | 5 | 6 ")
    print("\t\t\t--------------")
    print("\t\t\t 7 | 8 | 9 \n\n")

def declareWinner(whoseTurn):
    if whoseTurn == COMPUTER:
        print("COMPUTER has won")
    else:
        print("HUMAN has won")

def rowCrossed(board):
    for i in range(SIDE):
        if board[i][0] == board[i][1] == board[i][2] and board[i][0] != ' ':
            return True
    return False

def columnCrossed(board):
    for i in range(SIDE):
        if board[0][i] == board[1][i] == board[2][i] and board[0][i] != ' ':
            return True
    return False

def diagonalCrossed(board):
    if board[0][0] == board[1][1] == board[2][2] and board[0][0] != ' ':
        return True
    if board[0][2] == board[1][1] == board[2][0] and board[0][2] != ' ':
        return True
    return False

def gameOver(board):
    return rowCrossed(board) or columnCrossed(board) or diagonalCrossed(board)

# Updated take_turn accepting board and player
def take_turn(board, player):
    print("HUMAN's turn.")
    while True:
        position = input("Choose a position from 1-9: ")
        if position not in [str(i) for i in range(1, 10)]:
            print("Invalid input. Choose a position from 1-9: ")
            continue
        idx = int(position) - 1
        x, y = idx // SIDE, idx % SIDE
        if board[x][y] != " ":
            print("Position already taken. Choose a different position.")
            continue
        board[x][y] = player
        showBoard(board)
        break

def playTicTacToe(whoseTurn):
    board, moves = initialise()
    showInstructions()
    moveIndex = 0

    while not gameOver(board) and moveIndex < SIDE*SIDE:
        if whoseTurn == COMPUTER:
            while moveIndex < SIDE*SIDE:
                x = moves[moveIndex] // SIDE
                y = moves[moveIndex] % SIDE
                if board[x][y] == ' ':
                    board[x][y] = COMPUTERMOVE
                    print("COMPUTER has put a {} in cell {}".format(COMPUTERMOVE, moves[moveIndex]+1))
                    showBoard(board)
                    moveIndex += 1
                    break
                moveIndex += 1
            whoseTurn = HUMAN
        elif whoseTurn == HUMAN:
            take_turn(board, HUMANMOVE)
            moveIndex += 1
            whoseTurn = COMPUTER

    if not gameOver(board) and moveIndex >= SIDE*SIDE:
        print("It's a draw")
    else:
        whoseTurn = HUMAN if whoseTurn == COMPUTER else COMPUTER
        declareWinner(whoseTurn)

if __name__ == "__main__":
    playTicTacToe(COMPUTER)
```
